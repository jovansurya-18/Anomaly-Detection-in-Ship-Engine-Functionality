A low lubrication oil temperature may indicate it is not reaching its optimal operating temperature, potentially causing inadequate lubrication.
- **Coolant temperature:** High coolant temperature signals overheating, which various issues, including a failed thermostat, coolant leak, or insufficient coolant flow can cause. A low coolant temperature could suggest the engine is not reaching its optimal operating temperature, affecting performance and efficiency.

Issues with engines could lead to engine malfunctions, potential safety hazards, and downtime (e.g. delayed deliveries), resulting in the breakdown of a ship’s overall functionality, consequently impacting the business, such as affecting revenue via failure to deliver goods. # Anomaly detection with statistical methods

```python
# IQR methodology

columns = df.columns

for column in columns:
  sorted = df[column].sort_values()
  q1 = sorted.quantile(0.25)
  q3 = sorted.quantile(0.75)
  iqr = q3-q1
  lower_range = q1 - 1.5 * iqr
  upper_range = q3 + 1.5 * iqr
  df[f"{column}_anomaly"] = ((df[column] < lower_range) | (df[column] > upper_range)).astype(int)

df
```

```python
# finding the count of 0s(normal data points), 1s(anomalies), and the proportion (%) of anomalies for each feature

anomaly_columns = df.columns[6:]
# anomaly_columns
for anomaly_column in anomaly_columns:
  counts = df[anomaly_column].value_counts()
  anomaly_proportion = counts[1]/len(df[anomaly_column])*100
  print(f"{counts}")
  print(f"Anomaly proportion: {anomaly_proportion:.2f}%\n")
```

```python
# create new df with the 0s and 1s only
df_anomaly = df[df.columns[6:]]

# find the total number of anomalies for each sample
df_anomaly["total_1s"] = df_anomaly.sum(axis=1)

# assign the numerator - to find proportion in the for loop
numerators = df_anomaly["total_1s"].unique()

# assign the denominator
denominator = len(df_anomaly)

for numerator in numerators:
  sum_row_numerator = (df_anomaly["total_1s"] >= numerator).sum()
  proportion = sum_row_numerator/denominator*100
  print(f"If samples have at least {numerator} anomalies between features, then total outliers is {proportion:.2f}% - {sum_row_numerator} out of {denominator} rows")
```

Given six total features, setting the anomaly threshold to two anomalous features results in an anomaly rate of approximately 2.16%, which falls within the desired 1–5% range. In other words, any sample exhibiting anomalies in at least 2 of the 6 features is classified as an outlier, aligning the detection rate with the target threshold

# Detecting anomaly with unsupervised ML methods

One-class SVM

```python
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler

df_original = df[df.columns[:6]]

# standardise the data first before performing One-class SVM
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_original)

# using gamma = 0.5 and nu = 0.05 as the hyperparameter baseline
model = OneClassSVM(
    kernel = "rbf",
    gamma = 0.5,
    nu = 0.05 # allowing to accept 5% noise
)

model.fit(X_scaled)
```

```python
y_pred = model.predict(X_scaled)

df_scaled_svm = pd.DataFrame(X_scaled, columns = df.columns[:6])
df_scaled_svm["anomaly_svm"] = y_pred

df_scaled_svm
```

```python
df_scaled_svm["anomaly_svm"].value_counts()
```

```python
df_anomaly_svm = df_scaled_svm[df_scaled_svm["anomaly_svm"]==-1]
df_anomaly_svm
```

```python
anomaly_proportion = (
    df_anomaly_svm.shape[0]/
    df_scaled_svm.shape[0]
)

print(f"Proportion of anomalies using one-class SVM is: {anomaly_proportion*100:.2f}%")
```

Setting gamma = 0.5 and nu = 0.05 allows too much noise, resulting in an anomaly rate of 5.34%, which exceeds the desired 1–5% range. Let's tune the hyperparameters to reduce the number of detected anomalies.

```python
# set the model with different hyperparameter
model = OneClassSVM(
    kernel = "rbf",
    gamma = 0.3,
    nu = 0.03 # allowing to accept 3% noise
)

model.fit(X_scaled)

y_pred = model.predict(X_scaled)

df_scaled_svm = pd.DataFrame(X_scaled, columns = df.columns[:6])
df_scaled_svm["anomaly_svm"] = y_pred

df_anomaly_svm = df_scaled_svm[df_scaled_svm["anomaly_svm"]==-1]

anomaly_proportion = (
    df_anomaly_svm.shape[0]/
    df_scaled_svm.shape[0]
)

print(f"Proportion of anomalies using one-class SVM is: {anomaly_proportion*100:.2f}%")
```

Setting gamma = 0.3 and nu = 0.03 results in an anomaly rate of 3.02%, which falls within the desired 1–5% range. We will proceed with this configuration.

The next step is to visualise the data by first reducing the features to two components using PCA for dimensionality reduction.

# PCA with One-class SVM

```python
from sklearn.decomposition import PCA

df_scaled = pd.DataFrame(X_scaled, columns = df.columns[:6])

pca = PCA(n_components=2)
X_pca = pca.fit_transform(df_scaled)

X_pca.shape
```

```python
y_subset = df_scaled_svm["anomaly_svm"]

# function to plot the PCA
def plot_pca (X, title, y_subset):
  plt.figure(figsize=(10, 8))
  for i in np.unique(y_subset):
    plt.scatter(
        X[y_subset == i, 0],
        X[y_subset == i, 1],
        label=str(i))
  plt.xlabel('Component 1')
  plt.ylabel('Component 2')
  plt.legend()
  plt.title(title)
  plt.grid(True)
  plt.show()

plot_pca (X_pca, "PCA with One-class SVM", y_subset)
```

The PCA chart above illustrates the distribution and separation between normal data points (orange dots) and outliers (blue dots), where the outliers amount to 3.02% of the dataset

# Isolation forest

```python
# with isolation forest, we don't need to standardise the data like what we did with One-class SVM
# the standardisation will occur once we build the model and before we perform PCA for dimensionality reduction
from sklearn.ensemble import IsolationForest

df_iso = df[df.columns[:6]]

iso_forest = IsolationForest(
    n_estimators = 100,
    contamination = 0.05,
    random_state = 42
)

iso_forest.fit(df_iso)
```

```python
scores_pred = iso_forest.decision_function(df_iso)
scores_pred
```

```python
y_pred = iso_forest.predict(df_iso)
y_pred
```

```python
df_iso["anomaly_iso"] = y_pred
df_iso
```

```python
df_iso["anomaly_iso"].value_counts()
```

```python
df_iso_anomaly = df_iso[df_iso["anomaly_iso"]==-1]
df_iso_anomaly
```

```python
anomaly_proportion_iso = (
    df_iso_anomaly.shape[0]/
    df_iso.shape[0]
)

print(f"Proportion of anomaly using Isolation Forest is: {anomaly_proportion_iso*100:.3f}%")
```

Setting contamination = 0.5 allows too much noise, resulting in an anomaly rate of 5.001%, which exceeds the desired 1–5% range. Let's tune the hyperparameter to reduce the number of detected anomalies.

```python
df_iso = df[df.columns[:6]]

iso_forest = IsolationForest(
    n_estimators = 100,
    contamination = 0.03,
    random_state = 42
)

iso_forest.fit(df_iso)

y_pred = iso_forest.predict(df_iso)

df_iso["anomaly_iso"] = y_pred

df_iso_anomaly = df_iso[df_iso["anomaly_iso"]==-1]

anomaly_proportion_iso = (
    df_iso_anomaly.shape[0]/
    df_iso.shape[0]
)

print(f"Proportion of anomaly using Isolation Forest is: {anomaly_proportion_iso*100:.3f}%")
```

Setting contamination = 0.3 results in an anomaly rate of 3.005%, which falls within the desired 1–5% range. We will proceed with this configuration.

The next step is to visualise the data by first reducing the features to two components using PCA for dimensionality reduction.
